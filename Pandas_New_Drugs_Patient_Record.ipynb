```python
import pandas as pd
```

```python
# Import the dataset for the risk management of the new drugs by Health Canada.
risk_management_new_drug_health_canada = pd.read_csv("risk management_new drug_health canada.csv")
display(risk_management_new_drug_health_canada.head(61))
```

```python
# Remove any duplication from the dataset.
risk_management_new_drug_health_canada = risk_management_new_drug_health_canada.drop_duplicates()
display(risk_management_new_drug_health_canada.head(59))
```

```python
# Rename colums in the dataset.
risk_management_new_drug_health_canada = risk_management_new_drug_health_canada.rename({'Lead_Office_Fro': 'Lead_Office', 'Product_Line': 'Categories_Product', 'SAP_No': 'File_No', 'Review_Type': 'Department_Review_Type'})
display(risk_management_new_drug_health_canada.head(59))
```

```python
risk_management_new_drug_health_canada = risk_management_new_drug_health_canada.rename(columns={'Lead_Office_Fro': 'Lead_Office', 'Product_Line': 'Categories_Product', 'SAP_No': 'File_No', 'Review_Type': 'Department_Review_Type'})
display(risk_management_new_drug_health_canada.head(59))
```
